# Data visualization

```python
plt.figure(figsize=(10,5))
sns.set(style='dark')
sns.countplot(x='Do you typically check a daily weather report?',data=df,
              order= df['Do you typically check a daily weather report?'].value_counts().index,color='blue');
```

```python
fig = px.pie(df['Do you typically check a daily weather report?'].value_counts().reset_index(), 
             values='Do you typically check a daily weather report?', names='index')
fig.update_traces(textposition='inside', textinfo='percent+label')
fig.show("png")
```

```python
K=df[['Age','Do you typically check a daily weather report?']]
df1=df.groupby(['Age','Do you typically check a daily weather report?']).size().reset_index(name='Total Content')
fig = px.line(df1, x='Age', y='Total Content', color='Do you typically check a daily weather report?')
fig.show("png")
```

```python
K=df[['Age','What is your gender?']]
df1=df.groupby(['Age','What is your gender?']).size().reset_index(name='Total Content')
fig = px.line(df1, x='Age', y='Total Content', color='What is your gender?')
fig.show("png")
```

```python
sns.set(style='white')
plt.figure(figsize=(25,9))
sns.countplot(x='What is your gender?',data= df,
              hue= 'Do you typically check a daily weather report?',
              order = df['What is your gender?'].value_counts().index[0:10])
plt.show()
```

```python
plt.figure(figsize=(14,8))
plt.title('How likely or unlikely would you be to check the weather on that device?', fontsize=25)
sns.countplot(x="If you had a smartwatch (like the soon to be released Apple Watch), how likely or unlikely would you be to check the weather on that device?",
  data=df,
order= df['If you had a smartwatch (like the soon to be released Apple Watch), how likely or unlikely would you be to check the weather on that device?'].value_counts().index);
```

```python
K=df[['If you had a smartwatch (like the soon to be released Apple Watch), how likely or unlikely would you be to check the weather on that device?','Do you typically check a daily weather report?']]
df1=df.groupby(['If you had a smartwatch (like the soon to be released Apple Watch), how likely or unlikely would you be to check the weather on that device?','Do you typically check a daily weather report?']).size().reset_index(name='Total Content')
fig = px.scatter(df1, x='If you had a smartwatch (like the soon to be released Apple Watch), how likely or unlikely would you be to check the weather on that device?', 
                 y='Total Content', color='Do you typically check a daily weather report?')
fig.show("png")
```
