```python
#Հարաբերակցությունը ըստ գույների՝ հաղթող խմբում
White=len(df[df['winner']=='white'])
Black=len(df[df['winner']=='black'])
Draw = len(df[df['winner'] == 'draw'])
labels = ['White', 'Black', 'Draw']
colors = ['white','black', 'grey']
fig, ax = plt.subplots(figsize=(6, 6))

patches, texts, pcts = ax.pie([White, Black, Draw], labels=labels, autopct='%.2f%%', colors = colors, wedgeprops={'linewidth': 3.0, 'edgecolor': 'grey'}, textprops={'size': 'x-large'})
plt.setp(pcts[1], color='w', fontweight='bold')
ax.set_title('Winner colors by percents', fontsize=11);
```

Ակնհայտ է, որ բավականին ուժեղ կապ գոյություն ունի white_rating-ի և black_rating-ի միջև:

```python
#Ներկայացված է winner-ը ըստ opening_ply-ի:
plt.figure(figsize=(10,6))
df1=df[['winner','opening_ply']]
df2=df1.groupby(['opening_ply','winner']).size().reset_index(name='Total Content')
fig = px.line(df2, x='opening_ply', y='Total Content', color='winner',
              color_discrete_sequence=["black", "grey", "white"])
fig.show("png")
```

3-րդ քայում winner white-ը գերազանցում է winner black-ին, նորից ընդունում ենք 1-ին հիփոթեզը:

```python
df['rating_diff'] = df['white_rating'] - df['black_rating'] 
```

```python
plt.figure(figsize=(7,5))
winner_markers= [['white', 'o', 'darkgray'],
                ['black', '^', 'black']]
plt.figure(figsize=(16,9))
for winner, marker, color in winner_markers:
    sns.regplot(data = df.loc[df['winner']==winner], x = 'rating_diff', y = 'turns',
               x_jitter=5,fit_reg = False, marker = marker, scatter_kws={'color':color});

plt.legend(title='Winner',labels=['White','Black'],title_fontsize=12, fontsize=11);
plt.axvspan(0, 1100, facecolor='g', alpha=0.05)
plt.axvline(0, color='k')
plt.xlim(-1100,1100)
plt.xlabel('Rating Difference (White-Black)', fontsize=13)
plt.ylabel('Number of Turns', fontsize=13);
plt.title('Distribution of Winners', fontsize=14);
```

Այստեղ ներյակայացված է rating_diff-ը ըստ turns-ի քանակության:Ավելի բարձր Rating Difference (White-Black)-ի դեպքում White- քանակությունը գերզանցաում է Black-ին ՝ հաշվի առնելով turns- քանակը:

```python
#Անիմացիա է, որը արտացոլում է opening_ply-ի ավելացման դեպքում, ինչպես է փոխվում ըստ winner-ի victory_status-ը տրված turns-ի դեպքում:
fig = px.scatter(df, x="victory_status", y="turns", animation_frame="opening_ply", animation_group="winner"
                 , color="victory_status", hover_name="winner", facet_col="winner",
           )
fig.show("png")
```
