# Data Visualization


```python
cor = df.corr()
print(cor)
```

```python
plt.figure(figsize = (5,4))
plt.rc('xtick',labelsize =12)
plt.rc('ytick',labelsize =10)
sns.heatmap(cor.corr(),cmap ="YlGnBu")
```

```python
fig = px.scatter(df, x='consumption', y='co2_emmission', color='food_category')
fig.show()
```

```python
plt.ylabel('Consumption', size =15);
plt.xlabel('food category', size =15);
sns.barplot(x='food_category', y='consumption', data=df)
plt.xticks(rotation=80);
```

```python
sns.lineplot(data=df, x="food_category", y="co2_emmission")
plt.xticks(rotation=80);
```

```python
countries= ['Armenia', 'Russia', 'Belarus','Georgia', 'Canada',
       'Latvia','Ukraine', 'Turkey']
```

```python
sns.set_context("poster")
sns.set_style("darkgrid")
plt.rc('xtick', labelsize=40) 
plt.rc('ytick', labelsize=33)
plt.figure(figsize=(40,30))

for i,c in enumerate(countries):
    data2 = df[df.country == c]
    plt.subplot(4,2,i+1)
    sns.barplot(data= data2, x = 'co2_emmission', y ='food_category' )
    plt.xlabel('Installation in Millions',size = 20)
    plt.ylabel('')
    plt.title(c,size = 30)
    
plt.tight_layout()
plt.subplots_adjust(hspace= 0.4)
#mp.show()
```
