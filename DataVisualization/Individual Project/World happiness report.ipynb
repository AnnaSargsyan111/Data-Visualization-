####  Փոփոխակնների փոփոխությունը աշխարհում ըստ տարիների:

```python
def animated_graph(df1, column_name):
    fig = px.choropleth(df1.sort_values('year'), locations="Country name", color=column_name, 
                    locationmode='country names',
                    animation_frame='year',
                 color_continuous_midpoint=0.8,
                    color_continuous_scale=px.colors.sequential.dense,
#                         width=800, height=400
                   )
    fig.update_layout(title=column_name+' Comparison by country',autosize=True)
    fig.show('png')
    #fig.show()
    
```

```python
col = df1.loc[:, df1.dtypes == np.float64].columns
print(col)
```

```python
for cl in col:
    animated_graph(df1,cl)
```
